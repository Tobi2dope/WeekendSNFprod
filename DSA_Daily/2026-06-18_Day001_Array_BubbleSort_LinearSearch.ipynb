## PART 1 — Data Structure: Array (Python List)

An **array** is a contiguous block of memory that stores elements of the same (or mixed, in Python) type.
Python's built-in `list` is a **dynamic array** — it resizes automatically when it runs out of capacity.

### Memory layout
```
Index:  0       1       2       3       4
      ┌───────┬───────┬───────┬───────┬───────┐
      │  10   │  20   │  30   │  40   │  50   │
      └───────┴───────┴───────┴───────┴───────┘
```
- **Random access** is O(1) because index → memory address is a single arithmetic step.
- **Insertion/deletion in the middle** is O(n) because every element after the target must shift.

````python
# ─── Array: Core Operations ───────────────────────────────────────────────────

class Array:
    """
    A thin wrapper around Python's list that exposes common array operations
    with explicit complexity annotations.
    """

    def __init__(self, initial=None):
        self._data = list(initial) if initial else []

    # ── O(1) ──────────────────────────────────────────────────────────────────
    def get(self, index):
        """Read element at index — O(1)."""
        if not (0 <= index < len(self._data)):
            raise IndexError(f"Index {index} out of range (size={len(self._data)})")
        return self._data[index]

    def set(self, index, value):
        """Overwrite element at index — O(1)."""
        if not (0 <= index < len(self._data)):
            raise IndexError(f"Index {index} out of range")
        self._data[index] = value

    def append(self, value):
        """Add to the end — amortised O(1)."""
        self._data.append(value)

    # ── O(n) ──────────────────────────────────────────────────────────────────
    def insert(self, index, value):
        """Insert before index, shifting later elements right — O(n)."""
        self._data.insert(index, value)

    def delete(self, index):
        """Remove element at index, shifting later elements left — O(n)."""
        if not (0 <= index < len(self._data)):
            raise IndexError(f"Index {index} out of range")
        return self._data.pop(index)

    def __len__(self):
        return len(self._data)

    def __repr__(self):
        return f"Array({self._data})"


# ── Demo ──────────────────────────────────────────────────────────────────────
arr = Array([10, 20, 30, 40, 50])
print("Initial array:", arr)

print("\n--- O(1) operations ---")
print(f"  arr.get(2)          → {arr.get(2)}    (direct index access)")
arr.set(2, 99)
print(f"  arr.set(2, 99)      → {arr}")
arr.append(60)
print(f"  arr.append(60)      → {arr}")

print("\n--- O(n) operations ---")
arr.insert(1, 15)
print(f"  arr.insert(1, 15)   → {arr}  (elements 1+ shifted right)")
removed = arr.delete(1)
print(f"  arr.delete(1) → {removed}, result: {arr}  (elements 1+ shifted left)")

print("\n--- Slicing and iteration ---")
print(f"  First 3 elements : {arr._data[:3]}")
print(f"  Reversed         : {arr._data[::-1]}")
print(f"  Even-index items : {arr._data[::2]}")
````

## PART 2 — Sorting Algorithm: Bubble Sort

### Core Idea
On each **pass** through the array, compare every adjacent pair.  
If the left element is greater than the right, **swap** them.  
After one full pass, the **largest unsorted element has "bubbled up"** to its correct position.
Repeat until no swaps are needed.

### Step-by-step walkthrough on `[5, 3, 8, 1, 2]`
```
Pass 1:  [5,3,8,1,2] → swap(5,3) → [3,5,8,1,2]
         [3,5,8,1,2] → no swap   → [3,5,8,1,2]
         [3,5,8,1,2] → swap(8,1) → [3,5,1,8,2]
         [3,5,1,8,2] → swap(8,2) → [3,5,1,2,8]  ← 8 is now in place

Pass 2:  [3,5,1,2,8] → no swap   → [3,5,1,2,8]
         [3,5,1,2,8] → swap(5,1) → [3,1,5,2,8]
         [3,1,5,2,8] → swap(5,2) → [3,1,2,5,8]  ← 5 is now in place

Pass 3:  [3,1,2,5,8] → swap(3,1) → [1,3,2,5,8]
         [1,3,2,5,8] → swap(3,2) → [1,2,3,5,8]  ← 3 is now in place

Pass 4:  [1,2,3,5,8] → no swaps  → DONE ✓
```

### Complexity
| Case | Time | Space |
|------|------|-------|
| Best (already sorted) | O(n) | O(1) |
| Average | O(n²) | O(1) |
| Worst (reverse sorted) | O(n²) | O(1) |

````python
# ─── Bubble Sort ──────────────────────────────────────────────────────────────

def bubble_sort(arr, verbose=False):
    """
    Sort `arr` in-place using Bubble Sort.

    Optimisation: if a full pass completes with zero swaps,
    the array is already sorted — exit early (best-case O(n)).

    Args:
        arr     : list of comparable elements
        verbose : if True, print each swap so you can watch the algorithm work

    Returns:
        (sorted list, swap_count)
    """
    a = arr[:]          # work on a copy so the original is untouched
    n = len(a)
    swap_count = 0

    for pass_num in range(n - 1):
        swapped = False

        # After `pass_num` passes, the last `pass_num` elements are sorted.
        # We can safely shrink the inner loop's boundary each time.
        for j in range(n - 1 - pass_num):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]   # swap
                swap_count += 1
                swapped = True
                if verbose:
                    print(f"  Pass {pass_num+1}, swap indices [{j},{j+1}]: {a}")

        if not swapped:          # early-exit optimisation
            if verbose:
                print(f"  Pass {pass_num+1}: no swaps — early exit")
            break

    return a, swap_count
````

````python
# ─── Verbose walkthrough ──────────────────────────────────────────────────────
sample = [5, 3, 8, 1, 2]
print(f"Input : {sample}")
sorted_arr, swaps = bubble_sort(sample, verbose=True)
print(f"Output: {sorted_arr}  (total swaps: {swaps})")
````

````python
# ─── Test Cases ───────────────────────────────────────────────────────────────
test_cases = [
    ([5, 3, 8, 1, 2],        "random"),
    ([1, 2, 3, 4, 5],        "already sorted (best case)"),
    ([5, 4, 3, 2, 1],        "reverse sorted (worst case)"),
    ([42],                   "single element"),
    ([],                     "empty list"),
    ([3, 3, 1, 1, 2],        "with duplicates"),
    ([-4, 0, 7, -1, 3],      "with negatives"),
]

print(f"{'Input':<30} {'Sorted':<30} {'Swaps':<6} {'Case'}")
print("-" * 80)
for data, label in test_cases:
    result, swaps = bubble_sort(data)
    print(f"{str(data):<30} {str(result):<30} {swaps:<6} {label}")
````

## PART 3 — Searching Algorithm: Linear Search

### Core Idea
Scan every element from index 0 to the end.  
Return the index the moment the target is found; return `-1` if the full scan completes without a match.

```
Array : [10, 25, 37, 42, 55]
Target: 37

Step 1 → check index 0: 10 ≠ 37
Step 2 → check index 1: 25 ≠ 37
Step 3 → check index 2: 37 = 37 ✓  → return 2
```

### When to use Linear Search
- Array is **unsorted** (binary search requires sorted input)
- Array is **very small** (overhead of sorting isn't worth it)
- You need to find **all occurrences**, not just the first

### Complexity
| Case | Time | Space |
|------|------|-------|
| Best (target at index 0) | O(1) | O(1) |
| Average | O(n/2) → O(n) | O(1) |
| Worst (not present or last) | O(n) | O(1) |

````python
# ─── Linear Search ────────────────────────────────────────────────────────────

def linear_search(arr, target, find_all=False):
    """
    Search for `target` in `arr`.

    Args:
        arr      : list to search
        target   : value to find
        find_all : if True, return ALL indices where target appears;
                   if False, return the first index only (default)

    Returns:
        int  : first index found      (find_all=False, target present)
        list : all indices found      (find_all=True)
        -1   : target not in arr      (find_all=False, target absent)
        []   : target not in arr      (find_all=True,  target absent)
    """
    if find_all:
        results = []
        for i, element in enumerate(arr):
            if element == target:
                results.append(i)
        return results
    else:
        for i, element in enumerate(arr):
            if element == target:
                return i        # found — stop immediately
        return -1               # not found


# ─── Demo ─────────────────────────────────────────────────────────────────────
data = [10, 25, 37, 42, 55, 37, 99]
print("Array:", data)
print()

# First occurrence
idx = linear_search(data, 37)
print(f"linear_search(37)              → index {idx}  (value: {data[idx]})")

# All occurrences
all_idx = linear_search(data, 37, find_all=True)
print(f"linear_search(37, find_all)    → indices {all_idx}")

# Not found
missing = linear_search(data, 100)
print(f"linear_search(100)             → {missing}  (not found)")
````

````python
# ─── Test Cases ───────────────────────────────────────────────────────────────
search_tests = [
    ([10, 25, 37, 42, 55], 37,  "found in middle"),
    ([10, 25, 37, 42, 55], 10,  "found at start (best case)"),
    ([10, 25, 37, 42, 55], 55,  "found at end (worst case)"),
    ([10, 25, 37, 42, 55], 99,  "not in list"),
    ([],                   5,   "empty list"),
    ([7, 7, 7],            7,   "all duplicates — first only"),
    ([-3, 0, 3, 6],        0,   "find zero"),
]

print(f"{'Array':<28} {'Target':<8} {'Result':<8} {'Case'}")
print("-" * 70)
for arr, target, label in search_tests:
    result = linear_search(arr, target)
    found  = f"index {result}" if result != -1 else "not found"
    print(f"{str(arr):<28} {str(target):<8} {found:<14} {label}")
````

## PART 4 — Putting It All Together

A real-world mini-pipeline:
1. Store raw scores in an **Array**
2. Sort them with **Bubble Sort**
3. Find a specific score with **Linear Search**

````python
# ─── End-to-End Example ───────────────────────────────────────────────────────
import random
random.seed(42)

# 1. Build array of quiz scores
scores = Array([random.randint(50, 100) for _ in range(10)])
print("1. Raw scores  :", scores._data)

# 2. Sort with Bubble Sort
sorted_scores, n_swaps = bubble_sort(scores._data)
print(f"2. Sorted      : {sorted_scores}  ({n_swaps} swaps)")

# 3. Search for a passing threshold
PASS_SCORE = 75
pos = linear_search(sorted_scores, PASS_SCORE)
if pos != -1:
    print(f"3. Score {PASS_SCORE} found at index {pos} in sorted list")
else:
    print(f"3. Score {PASS_SCORE} not present; first score ≥ {PASS_SCORE}: ", end="")
    first_passing = next((s for s in sorted_scores if s >= PASS_SCORE), None)
    print(first_passing)

# 4. Summary stats
print(f"\n--- Summary ---")
print(f"   Min  : {sorted_scores[0]}")
print(f"   Max  : {sorted_scores[-1]}")
print(f"   Mean : {sum(sorted_scores)/len(sorted_scores):.1f}")
passing = [s for s in sorted_scores if s >= PASS_SCORE]
print(f"   Passing (≥{PASS_SCORE}): {len(passing)}/{len(sorted_scores)}  → {passing}")
````

## Complexity Cheat Sheet

```
┌─────────────────┬──────────┬──────────┬──────────┬─────────┐
│ Operation       │ Best     │ Average  │ Worst    │ Space   │
├─────────────────┼──────────┼──────────┼──────────┼─────────┤
│ Array Access    │ O(1)     │ O(1)     │ O(1)     │ O(n)    │
│ Array Insert    │ O(1)*    │ O(n)     │ O(n)     │ O(1)    │
│ Array Delete    │ O(1)*    │ O(n)     │ O(n)     │ O(1)    │
├─────────────────┼──────────┼──────────┼──────────┼─────────┤
│ Bubble Sort     │ O(n)     │ O(n²)    │ O(n²)    │ O(1)    │
├─────────────────┼──────────┼──────────┼──────────┼─────────┤
│ Linear Search   │ O(1)     │ O(n)     │ O(n)     │ O(1)    │
└─────────────────┴──────────┴──────────┴──────────┴─────────┘
 * insert/delete at end
```

## Key Takeaways
- Arrays give you **O(1) random access** — the killer feature.
- Bubble Sort is simple but slow — fine for **≤1 000 elements**, impractical beyond that.
- Linear Search is the only option on **unsorted data** — consider Binary Search once the array is sorted.
